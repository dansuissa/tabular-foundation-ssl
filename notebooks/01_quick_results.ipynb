# Quick benchmark results

Load aggregated CSV outputs and inspect method performance across datasets.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

PROJECT_ROOT = Path('..').resolve()
SUMMARY_PATH = PROJECT_ROOT / 'results' / 'aggregated' / 'summary_by_method.csv'
FAILURES_PATH = PROJECT_ROOT / 'results' / 'aggregated' / 'failures.csv'

if not SUMMARY_PATH.exists():
    raise FileNotFoundError(
        'No aggregated summary found. Run:\n'
        '  python -m src.run_benchmark\n'
        '  python -m src.aggregate_results'
    )

summary = pd.read_csv(SUMMARY_PATH)
summary.head()
```

```python
metric_cols = [col for col in summary.columns if col.startswith('metric_') and col.endswith('_mean')]
if metric_cols:
    plot_df = summary.copy()
    plot_df['method_dataset'] = plot_df['method'] + ' | ' + plot_df['dataset']
    plt.figure(figsize=(10, 5))
    sns.barplot(data=plot_df, x='method_dataset', y=metric_cols[0])
    plt.xticks(rotation=45, ha='right')
    plt.title(f'Top-line comparison ({metric_cols[0]})')
    plt.tight_layout()
    plt.show()
else:
    print('No metric summary columns available yet.')
```

```python
if FAILURES_PATH.exists():
    failures = pd.read_csv(FAILURES_PATH)
    failures
else:
    print('No failures logged.')
```
